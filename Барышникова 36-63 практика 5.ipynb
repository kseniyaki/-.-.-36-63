```python
import json

def read_file(filename):
    try:
        with open(filename,'r',encoding='utf-8') as file:
            students=json.load(file)
        return students
    except FileNotFoundError:
        print(f'Ошибка: файл {filename} не найден')
        return None
    except json.JSONDecodeError:
        print(f'Ошибка: файл {filename} поврежден или является на коректным JSON')
        return None

def write_file(filename,data):
    try:
        with open(filename,'w',encoding='utf-8') as file:
            json.dump(data,file,ensure_ascii=False,indent=4)
        print(f'Отчет сохраннен в файл {filename}')
    except PermissionError:
        print(f'Ошибка: нет прав для записи в файл {filename}')

students=read_file('students.json')

if students is not None:
    report_data=[]
    total_sum=0
    total_count=0

    try:
        for student in students:
            try:
                name=student['name']
                grades=student['grades']
                average=sum(grades)/len(grades)

                report_data.append({
                    'name': name,
                    'average': round(average,2)
                })

                total_sum+=sum(grades)
                total_count+=len(grades)
            except KeyError as e:
                print(f'Ошибка: отсутсвует ключ {e}. Запись пропущена.')
            except TypeError:
                print(f'Ошибка: неверный формат оценок. Запись пропущена.')
            except ZeroDivisionError:
                print(f'Ошибка: список оценок пуст. Запись пропущена.')
    except TypeError:
        print(f'Ошибка: неверный формат списка студентов')

    if total_count>0:
        overall_avg=total_sum/total_count
    else:
        overall_avg=0

    final_data={
        'report': report_data,
        'overall_average': round(overall_avg,2)
    }
    write_file('report.json',final_data)
```
