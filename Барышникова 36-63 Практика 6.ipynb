```python
class Student:
    def __init__(self, name, group,tel_number, grades=None):
        self.name = name
        self.group = group
        self._tel_number = tel_number
        self.grades = grades if grades is not None else []

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        if not self.grades:
            return 0
        return sum(self.grades)/len(self.grades)
    def info(self):
        average=self.get_average()
        return f'Студент: {self.name}, Группа: {self.group}, Тел.: {self._tel_number}, Средний балл: {average: .2f}'

student1=Student('Ксения','36-63','89869251111')
student2=Student('София','26-42','89171347653',[5,4,4,3])
student3=Student('Никита','67-52','89523467867')

student1.add_grade(5)
student1.add_grade(4)
student1.add_grade(5)

student2.add_grade(5)

student3.add_grade(3)
student3.add_grade(5)
student3.add_grade(4)

print(student1.info())
print(student2.info())
print(student3.info())
```
