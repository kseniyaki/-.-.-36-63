```python
#2 практика
#задание 1
employed=bool(int(input('Трудоустроен (да-1,нет-0): ')))
vacation=bool(int(input('В отпуске (да-1,нет-0): ')))
res=employed and not vacation
print('Установить будильник, ',res)
```

```python
#задание 2
operation=str(input('Выберите операцию (+,-,*,/): '))
value_1=int(input('Введите первое число: '))
value_2=int(input('Введите второе число: '))
if operation=='+':
    res=value_1+value_2
    print(f'{value_1} + {value_2} = ',res)
elif operation=='-':
    res=value_1-value_2
    print(f'{value_1} - {value_2} = ',res)
elif operation=='*':
    res=value_1*value_2
    print(f'{value_1} * {value_2} = ',res)
else:
    res=value_1/value_2
    print(f'{value_1} / {value_2} = ',res)
```

```python
#задание 3
a=int(input('Введите первое число: '))
b=int(input('Введите второе число: '))
sum=0
for i in range(a,b+1):
    sum+=i
print(sum)
```

```python
#задание 4
m=int(input('Введите номер месяца: '))
if m>=10: print('Четвертый квартал')
elif m>=7: print('Третий квартал')
elif m>=4: print('Второй квартал')
else: print('Первый квартал')
```

```python
#задание 5
k=0
sum=0
cur=0
while cur>=0:
    a=int(input('Введите балл (от 0 до 100, чтобы прекратить ввод -1): '))
    if a==-1:break
    sum+=a
    k+=1
sr=sum/k
if sr>=90: print('Оценка A')
elif sr>=80: print('Оценка B')
elif sr>=70: print('Оценка C')
elif sr>=60: print('Оценка D')
else: print('Оценка F')
```
