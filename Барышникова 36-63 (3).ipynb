```python
#Практическая работа 4
#Задание 1
def f(v1,v2,oper):
    if oper=='+':
        return v1+v2
    elif oper=='-':
        return v1-v2
    elif oper=='*':
        return v1*v2
    else:
        if v2!=0:
            return v1/v2
        else:
            return 'Деление на ноль'
oper=str(input('Выберите операцию (+,-,*,/): '))
v1=int(input('Введите первое число: '))
v2=int(input('Введите второе число: '))
print(f(v1,v2,oper))

def g(m):
    if m in [1,2,3]:
        k='Первый'
        s='Зима' if m in [1,2] else 'Весна'
    elif m in [4,5,6]:
        k='Второй'
        s='Весна' if m in [4,5] else 'Лето'
    elif m in [7,8,9]:
        k='Третий'
        s='Лето' if m in [7,8] else 'Осень'
    else:
        k='Четвертый'
        s='Осень' if m in [10,11] else 'Зима'
    return f'{k} квартал, Сезон:{s}'
m=int(input('Введите номер месяца: '))
print(g(m))
```

```python
#Задание 2
def f(s):
    count={}
    for k in s:
        count[k]=count.get(k,0)+1
    return count
s=str(input('Введите предложение: '))
print(f(s))
```

```python
#Задание 3
def f(s,st,e):
    k=0
    for i in s:
        if st<=i<=e:
            k+=1
    return k
s='abcdifghijklmnopqrstuvwxyz'
print(f'Строка: {s}')
print(f(s,'m','z'))
```
