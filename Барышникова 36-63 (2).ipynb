```python
#Задание 2
list=['банан','яблоко','апельсин','виноград','груша','яблоко']
list_new=[]
for i in list:
    if i not in list_new:
        list_new.append(i)
    else:
        continue
print(list_new)
```

```python
#Задание 3
list=['банан','яблоко','апельсин','виноград','груша','яблоко','груша','груша']
alf={}
for i in list:
    if i in alf:
        alf[i] +=1
    else:
        alf[i]=1
print(alf)
```
